```python
import pandas as pd
import numpy as np
from glob import glob

import cv2
import matplotlib.pylab as plt
```

```python
#converting matplot to cv2
dog_plt_rgb=cv2.cvtColor(dog_img,cv2.COLOR_RGB2BGR)
fig,ax=plt.subplots()
ax.imshow(dog_plt_rgb)
ax.axis('off')
plt.show()
```

```python
dog_gray=cv2.cvtColor(dog_img,cv2.COLOR_RGB2GRAY)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(dog_gray,cmap='Greys')
ax.axis('off')
ax.set_title('Gray image')
plt.show()
```

```python
dog_resized=cv2.resize(dog_img,None,fx=0.25,fy=0.25)
fig,axs=plt.subplots(figsize=(8,8))
axs.imshow(dog_resized)
axs.axis('off')
axs.set_title('Resized Image (Scale)')
plt.show()
```

```python
dog_resize=cv2.resize(dog_img,(100,200))
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(dog_resize)
ax.axis('off')
ax.set_title('Resized Image (Fixed Size)')
plt.show()
```

```python
dog_extended=cv2.resize(dog_img,(5000,5000),interpolation=cv2.INTER_CUBIC)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(dog_extended)
ax.axis('off')
ax.set_title('Scaled Image')
plt.show()
```

```python
dog_extended=cv2.resize(dog_img,(5000,5000),interpolation=cv2.INTER_LINEAR)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(dog_extended)
ax.axis('off')
ax.set_title('Scaled Image')
plt.show()
```

```python
dog_extended=cv2.resize(dog_img,(5000,5000),interpolation=cv2.INTER_NEAREST)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(dog_extended)
ax.axis('off')
ax.set_title('Scaled Image')
# dog_extended.shape
plt.show()
```

```python
kernel_sharpening = np.array([[-1,-1,-1],
                            [-1,9,-1],
                            [-1,-1,-1]])
sharpened=cv2.filter2D(dog_img,-1,kernel_sharpening)

fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(sharpened)
ax.axis('off')
ax.set_title('Sharpened Image')
plt.show()
```

```python
sharpened.shape
```

```python
blur_kernel=np.ones((3,3),np.float32) /9
blurred=cv2.filter2D(dog_img,-1,blur_kernel)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(blurred)
ax.axis('off')
ax.set_title('Blurred Image')
plt.show()
```

```python
blurred.shape
```

```python
blur_kernel=np.ones((3,3),np.float32) /3
blurred=cv2.filter2D(dog_img,-1,blur_kernel)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(blurred)
ax.axis('off')
ax.set_title('Blurred Image')
plt.show()
```

```python
blur_kernel=np.ones((3,3),np.float32) /100
blurred=cv2.filter2D(dog_img,-1,blur_kernel)
fig,ax=plt.subplots(figsize=(8,8))
ax.imshow(blurred)
ax.axis('off')
ax.set_title('Blurred Image')
plt.show()
```

```python
plt.imsave('mpl_dog.png',blurred)
cv2.imwrite('cv2_dog.png',blurred)
```

### Histogram Equalization & Binary Inversion
Applying concepts from the page_2 notes.

```python
dog_gray_eq = cv2.equalizeHist(dog_gray)
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].imshow(dog_gray, cmap='gray')
axs[0].set_title('Original Grayscale')
axs[0].axis('off')
axs[1].imshow(dog_gray_eq, cmap='gray')
axs[1].set_title('Histogram Equalized')
axs[1].axis('off')
plt.show()
```

```python
# Binary inversion
_, dog_binary_inv = cv2.threshold(dog_gray, 127, 255, cv2.THRESH_BINARY_INV)
fig, ax = plt.subplots(figsize=(6, 6))
ax.imshow(dog_binary_inv, cmap='gray')
ax.set_title('Binary Inverted')
ax.axis('off')
plt.show()
```
